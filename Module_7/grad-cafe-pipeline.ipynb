```python
import boto3
import json

bucket_name = 'gradcafe-js-2026'
file_key = 'applicant_data.json'
local_path = 'applicant_data_SM.json'

s3 = boto3.client('s3')

# Download the file
s3.download_file(bucket_name, file_key, local_path)
print(f"✅ Downloaded {file_key} from S3 bucket {bucket_name}")

# Verify
with open(local_path, 'r') as f:
    data = json.load(f)
    print(f"📊 Loaded {len(data.get('results', []))} records")
print("✅ Data saved as 'applicant_data_SM.json'")
```
